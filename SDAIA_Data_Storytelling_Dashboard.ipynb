```python
import pandas as pd         #استيراد مكتبه مسؤوله عن البيانات والجداول
import plotly.express as px     #مسؤوله عن الرسومات البيانيه التفاعليه
```

```python


# 1. قراءة الملف
df = pd.read_csv('tayseer_services.csv')

# 2. رسم بياني تفاعلي يربط فئة الخدمة بعدد المعاملات مفرزاً حسب المنطقة
fig = px.bar(
    df,                                #مصدر البيانات الاساسي
    x='service_category',              # المحورالافقي ( الخدمات )
    y='transactions',                  # العمودي ( اعداد  المعاملات )
    color='region',                    # تلوين الاعمده حسب المنطقه
    barmode='group',                   # جعل الاعمده جانب بعض للمقارنه
    title="تحليل المعاملات حسب فئة الخدمة والمنطقة"             #العنوان الرئيسي للرسم
)

# 3. عرض الرسم البياني
fig.show()
```


1. **الملاحظة الرئيسية:** تختلف أعداد المعاملات بشكل ملحوظ بين فئات الخدمات المختلفة، حيث تستحوذ بعض الخدمات على النصيب الأكبر من تفاعل المستخدمين.


2. **فرصة التحسين:** الفئات الأكثر نشاطاً تمثل نقطة تركيز عالية لمستفيدي الخدمة، مما يتطلب ضمان استمرارية كفاءتها.


3. **التوصية للقرار:** توجيه الموارد والدعم الفني للخدمات الأكثر طلباً لرفع كفاءة الأداء، ودراسة أسباب تفاوت الاستخدام بين المناطق المختلفة.
